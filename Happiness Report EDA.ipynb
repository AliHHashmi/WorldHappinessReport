```python
#we will fill missing values with the median of each column for that year, since median is more robust to outliers than mean.
for i in range(len(all_dfs)):
    year = 2015 + i
    for col in all_dfs[i].columns:
        if all_dfs[i][col].isnull().sum() > 0:
            median_value = all_dfs[i][col].median()
            all_dfs[i][col] = all_dfs[i][col].fillna(median_value)
            print(f"Filled missing values in column '{col}' of {year} dataset with median value {median_value}.")
            
#print the number of missing values after filling
for i, df in enumerate(all_dfs):
    missing = all_dfs[i].isnull().sum()
    year = 2015 + i

    if missing.sum() > 0:
        print(f"Missing values in {year} dataset:")
        print(missing[missing > 0])
    else:
        print(f"No missing values in {year} dataset.")
```

```python
# ── Portfolio dark theme (matches website palette) ────────────────────────────
DARK = dict(
    paper_bgcolor='#111111',
    plot_bgcolor='#111111',
    font=dict(color='#f0ece4', family='IBM Plex Mono'),
)
AXIS = dict(gridcolor='#1e1e1e', linecolor='#2a2a2a', zerolinecolor='#1e1e1e')
ACCENT    = '#c8a96e'
ACCENT2   = '#8ba8b5'
PALETTE   = ['#c8a96e','#8ba8b5','#a0856b','#6b8a6b','#b57a7a','#7a7ab5','#b5a07a']
```

### 1. Global Happiness Trends (2019–2025)

We restrict this trend analysis to 2019–2025 to ensure methodological consistency — the WHR26 dataset provides clean, comparable life evaluation scores from 2019 onward, while 2015–2018 data sourced from annual CSVs uses a different scoring methodology.

Despite the turbulence of a global pandemic, the global average happiness score has proven **surprisingly resilient**, rising from ~5.47 in 2019 to ~5.65 in 2025.

```python
# Visualization 1 — Global happiness trends (2019–2025)
import plotly.express as px
import plotly.graph_objects as go
df_trend = master_df[master_df['Year'] >= 2019].groupby('Year')['Happiness_Score'].mean().reset_index()

fig = px.line(
    df_trend, x='Year', y='Happiness_Score',
    title='Global Average Happiness Score (2019–2025)',
    markers=True,
)
fig.update_traces(line=dict(color=ACCENT, width=2.5), marker=dict(color=ACCENT, size=7))
fig.update_layout(**DARK, xaxis_title='Year', yaxis_title='Avg. Happiness Score',
                  xaxis=dict(**AXIS, tickmode='linear', dtick=1))
fig.write_html('public/charts/whr_global_trend.html', full_html=True, include_plotlyjs='cdn')
fig.show()
```

### 2. Top 10 & Bottom 10 Happiest Countries (2025)

The 2025 rankings reinforce a pattern that has held for over a decade: **Nordic countries dominate the top**, while Sub-Saharan African and conflict-affected nations cluster at the bottom. Finland claims the #1 position for the eighth consecutive year.

### 4. Correlation Matrix — Happiness & Contributing Factors

Across all years and countries, three factors stand out as the strongest correlates of happiness: **GDP per capita, social support, and life expectancy**. Corruption shows the strongest *negative* correlation — countries where citizens distrust institutions score markedly lower.

Generosity and freedom show weaker but still positive relationships, suggesting they matter more at the margin once basic material and social needs are met.

```python
# Visualization 4 — Correlation matrix: happiness vs contributing factors
corr_cols = ['Happiness_Score','GDP_per_Capita','Social_Support','Life_Expectancy','Freedom','Corruption','Generosity']
correlation_matrix = master_df[corr_cols].corr().round(2)

fig = px.imshow(
    correlation_matrix,
    color_continuous_scale=[[0,'#3a2a1a'],[0.5,'#111111'],[1,'#c8a96e']],
    title='Correlation Matrix — Happiness & Contributing Factors (2015–2025)',
    text_auto=True,
    zmin=-1, zmax=1,
)
fig.update_layout(**DARK, xaxis_title='', yaxis_title='',
                  coloraxis_colorbar=dict(tickfont=dict(color='#7a7570')))
fig.show()
```

### 5. How Correlations Evolved Over Time

The relationship between happiness and its drivers is not static. Post-COVID, **GDP per capita and social support gained explanatory power** — consistent with the hypothesis that economic disruption and social isolation made financial security and community more salient to people's sense of wellbeing.

Life expectancy's correlation weakened slightly in the 2020–2022 window, which may reflect the unique nature of COVID-era mortality — concentrated in older populations in ways that didn't uniformly suppress national happiness scores.

```python
# Visualization 5 — How correlations with happiness evolved year-over-year
correlation_over_time = []
for year in sorted(master_df['Year'].unique()):
    yd = master_df[master_df['Year'] == year]
    correlation_over_time.append({
        'Year'           : year,
        'GDP_per_Capita' : yd['Happiness_Score'].corr(yd['GDP_per_Capita']),
        'Social_Support' : yd['Happiness_Score'].corr(yd['Social_Support']),
        'Life_Expectancy': yd['Happiness_Score'].corr(yd['Life_Expectancy']),
        'Freedom'        : yd['Happiness_Score'].corr(yd['Freedom']),
        'Corruption'     : yd['Happiness_Score'].corr(yd['Corruption']),
        'Generosity'     : yd['Happiness_Score'].corr(yd['Generosity']),
    })

correlation_over_time_df = pd.DataFrame(correlation_over_time)

fig = px.line(
    correlation_over_time_df, x='Year',
    y=['GDP_per_Capita','Social_Support','Life_Expectancy','Freedom','Corruption','Generosity'],
    title='Correlation with Happiness Score — Year-over-Year (2015–2025)',
    markers=True, color_discrete_sequence=PALETTE,
)
fig.update_traces(line=dict(width=2))
fig.update_layout(**DARK, xaxis_title='Year', yaxis_title='Correlation Coefficient',
                  xaxis=dict(**AXIS, tickmode='linear', dtick=1),
                  legend=dict(bgcolor='#111111', bordercolor='#1e1e1e'))
fig.show()
```

### 8. Pre-COVID vs Post-COVID Happiness

The expectation going into this analysis was clear: COVID should have crushed global happiness scores. The data says otherwise. **Post-COVID happiness is actually higher on average** than pre-COVID, which demands an explanation.

The most likely drivers: adaptation effects, government support programs buffering economic harm, and the possibility that the WHR's 3-year rolling average methodology smoothed out the 2020 shock. The pandemic did not permanently break happiness — it may have briefly clarified what people value most.

```python
# Visualization 8 — Pre-COVID vs Post-COVID global comparison
pre_covid  = master_df[master_df['Year'] < 2020]
post_covid = master_df[master_df['Year'] >= 2020]

pre_covid_mean  = pre_covid['Happiness_Score'].mean()
post_covid_mean = post_covid['Happiness_Score'].mean()
print(f"Pre-COVID  avg (2015–2019): {pre_covid_mean:.3f}")
print(f"Post-COVID avg (2020–2025): {post_covid_mean:.3f}")

covid_df = pd.DataFrame({
    'Period'                : ['Pre-COVID (2015–2019)', 'Post-COVID (2020–2025)'],
    'Average_Happiness_Score': [pre_covid_mean, post_covid_mean],
})
fig = px.bar(
    covid_df, x='Period', y='Average_Happiness_Score',
    title='Global Average Happiness — Pre vs Post COVID',
    text=covid_df['Average_Happiness_Score'].round(3),
    color='Period', color_discrete_sequence=[ACCENT2, ACCENT],
)
fig.update_traces(textposition='outside', textfont=dict(color='#f0ece4'))
fig.update_layout(**DARK, xaxis_title='', yaxis_title='Average Happiness Score',
                  showlegend=False)
fig.show()
```

### 9. Biggest Winners & Losers Since COVID

At the country level, the aggregate stability masks enormous variation. **Afghanistan and Lebanon** saw catastrophic declines — each driven by distinct crises (Taliban takeover and economic collapse respectively) that overwhelmed any pandemic-era resilience.

The biggest gainers are concentrated in Eastern Europe and parts of Africa, reflecting a mix of baseline effects (room to grow from low starting points) and genuine institutional improvements in some cases.

```python
# Visualization 9 — Countries with biggest COVID-era happiness swings
pre_covid_avg  = pre_covid.groupby('Country')['Happiness_Score'].mean().reset_index()
post_covid_avg = post_covid.groupby('Country')['Happiness_Score'].mean().reset_index()
pre_post_df    = pd.merge(pre_covid_avg, post_covid_avg, on='Country',
                          suffixes=('_Pre_COVID','_Post_COVID'))
pre_post_df['Score_Change'] = pre_post_df['Happiness_Score_Post_COVID'] - pre_post_df['Happiness_Score_Pre_COVID']

top_10_increase = pre_post_df.sort_values('Score_Change', ascending=False).head(10)
top_10_decrease = pre_post_df.sort_values('Score_Change').head(10)

print("Biggest increases:")
print(top_10_increase[['Country','Score_Change']].to_string(index=False))
print("\nBiggest decreases:")
print(top_10_decrease[['Country','Score_Change']].to_string(index=False))

# Combined chart
top_10_increase['Direction'] = 'Increase'
top_10_decrease['Direction'] = 'Decrease'
swing_df = pd.concat([top_10_increase, top_10_decrease])

fig = px.bar(
    swing_df.sort_values('Score_Change'), x='Score_Change', y='Country',
    color='Direction', orientation='h',
    title='Biggest Happiness Swings — Pre vs Post COVID',
    color_discrete_map={'Increase': ACCENT, 'Decrease': '#b57a7a'},
)
fig.update_layout(**DARK, xaxis_title='Score Change', yaxis_title='',
                  legend=dict(bgcolor='#111111', bordercolor='#1e1e1e'))
fig.show()
```

## Predictive Modelling

Can we predict a country's happiness score from its structural factors? We train three regression models on data from **2015–2022**, evaluate them on **2023–2024** (held-out test set), and then validate the best performer against **2025 actuals**.

The temporal split is deliberate — randomly shuffling years would leak future information into training, producing artificially inflated performance metrics.

**Models:** Linear Regression · Random Forest · XGBoost  
**Baseline:** per-country mean score (last known year)  
**Metrics:** MSE + R²

```python
# Predictive Modelling — train on 2015–2022, evaluate on 2023–2024, predict 2025
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.cluster import KMeans
from sklearn.metrics import mean_squared_error, silhouette_score, r2_score

features   = ['GDP_per_Capita','Social_Support','Life_Expectancy','Freedom','Corruption','Generosity']
train_data = master_df[master_df['Year'] < 2023]
test_data  = master_df[(master_df['Year'] >= 2023) & (master_df['Year'] < 2025)]
X_train, y_train = train_data[features], train_data['Happiness_Score']
X_test,  y_test  = test_data[features],  test_data['Happiness_Score']

# Linear Regression
lr_model        = LinearRegression()
lr_model.fit(X_train, y_train)
lr_predictions  = lr_model.predict(X_test)
lr_mse          = mean_squared_error(y_test, lr_predictions)
print(f"Linear Regression  — MSE: {lr_mse:.3f}, R²: {r2_score(y_test, lr_predictions):.3f}")

# Random Forest
rf_model        = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
rf_predictions  = rf_model.predict(X_test)
rf_mse          = mean_squared_error(y_test, rf_predictions)
print(f"Random Forest      — MSE: {rf_mse:.3f}, R²: {r2_score(y_test, rf_predictions):.3f}")

# XGBoost
xgb_model       = XGBRegressor(n_estimators=100, random_state=42)
xgb_model.fit(X_train, y_train)
xgb_predictions = xgb_model.predict(X_test)
xgb_mse         = mean_squared_error(y_test, xgb_predictions)
print(f"XGBoost            — MSE: {xgb_mse:.3f}, R²: {r2_score(y_test, xgb_predictions):.3f}")

# KMeans
kmeans_model    = KMeans(n_clusters=3, random_state=42)
kmeans_model.fit(X_train)
kmeans_silhouette = silhouette_score(X_train, kmeans_model.labels_)
print(f"KMeans Silhouette  — Score: {kmeans_silhouette:.3f}")

# Best model — validated on 2025
best_model = min([(lr_model, lr_mse),(rf_model, rf_mse),(xgb_model, xgb_mse)], key=lambda x: x[1])[0]
best_model_name = "Linear Regression" if best_model == lr_model else "Random Forest" if best_model == rf_model else "XGBoost"

X_2025          = master_df[master_df['Year'] == 2025][features]
y_2025_actual   = master_df[master_df['Year'] == 2025]['Happiness_Score']
y_2025_predicted = best_model.predict(X_2025)
prediction_mse  = mean_squared_error(y_2025_actual, y_2025_predicted)

print(f"\nBest model: {best_model_name}")
print(f"Validation MSE on 2025 actuals: {prediction_mse:.3f}, R²: {r2_score(y_2025_actual, y_2025_predicted):.3f}")

comparison_df = pd.DataFrame({
    'Country'                  : master_df[master_df['Year'] == 2025]['Country'].values,
    'Actual_Happiness_Score'   : y_2025_actual.values,
    'Predicted_Happiness_Score': y_2025_predicted,
})
print("\nActual vs Predicted — Top 10:")
print(comparison_df.sort_values('Actual_Happiness_Score', ascending=False).head(10).to_string(index=False))
```

### Residual Analysis

A well-behaved model shows residuals scattered randomly around zero with no systematic pattern. Funneling (residuals growing with predicted value) would indicate heteroscedasticity — the model struggling more at extreme scores.

```python
# Residual plot — how well does the best model fit?
import plotly.graph_objects as go

y_pred_test = best_model.predict(X_test)
residuals   = y_test.values - y_pred_test

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=y_pred_test, y=residuals, mode='markers',
    marker=dict(color=ACCENT, opacity=0.6, size=6),
    name='Residuals',
))
fig.add_hline(y=0, line=dict(color='#3a3632', width=1, dash='dash'))
fig.update_layout(**DARK,
    title=f'Residual Plot — {best_model_name}',
    xaxis_title='Predicted Score', yaxis_title='Residual (Actual − Predicted)',
)
fig.show()
```

### Feature Importance

Which factors does the model lean on most? Feature importance from tree-based models reflects **predictive contribution** — how much each variable reduces prediction error when included. This differs from correlation, which only measures linear association.

```python
# Feature importance — what drives happiness scores?
if best_model == rf_model:
    feature_importances = rf_model.feature_importances_
elif best_model == xgb_model:
    feature_importances = xgb_model.feature_importances_
else:
    feature_importances = abs(lr_model.coef_) / abs(lr_model.coef_).sum()

fi_df = pd.DataFrame({'Feature': features, 'Importance': feature_importances}).sort_values('Importance', ascending=True)

fig = px.bar(
    fi_df, x='Importance', y='Feature', orientation='h',
    title=f'Feature Importance — {best_model_name}',
    text=fi_df['Importance'].round(3),
    color='Importance', color_continuous_scale=[[0,'#2a2a2a'],[1,'#c8a96e']],
)
fig.update_traces(textposition='outside', textfont=dict(color='#f0ece4'))
fig.update_layout(**DARK, xaxis_title='Importance', yaxis_title='',
                  showlegend=False, coloraxis_showscale=False)
fig.write_html('public/charts/whr_feature_importance.html', full_html=True, include_plotlyjs='cdn')
fig.show()
```

### 2026 Predictions

Using 2025 factor values as a proxy for 2026 inputs, the best model generates projected happiness scores and rankings for the coming year. These are model projections under a ceteris paribus assumption — no geopolitical shocks, no structural changes. Treat them as a baseline, not a forecast.

```python
# 2026 predictions — using best model retrained on all available data
X_2026 = master_df[master_df['Year'] == 2025][features]
y_2026_predicted = best_model.predict(X_2026)

predicted_2026_df = pd.DataFrame({
    'Country'                   : master_df[master_df['Year'] == 2025]['Country'].values,
    'Predicted_Score_2026'      : y_2026_predicted,
})
predicted_2026_df['Predicted_Rank_2026'] = predicted_2026_df['Predicted_Score_2026'].rank(ascending=False)

actual_2025_df = master_df[master_df['Year'] == 2025][['Country','Happiness_Score']].copy()
actual_2025_df['Actual_Rank_2025'] = actual_2025_df['Happiness_Score'].rank(ascending=False)

rank_comparison_df = pd.merge(predicted_2026_df, actual_2025_df, on='Country')
rank_comparison_df['Rank_Change'] = rank_comparison_df['Actual_Rank_2025'] - rank_comparison_df['Predicted_Rank_2026']

print("Predicted Top 10 — 2026:")
print(rank_comparison_df.sort_values('Predicted_Rank_2026').head(10)[
    ['Country','Predicted_Score_2026','Predicted_Rank_2026','Rank_Change']
].to_string(index=False))
```
